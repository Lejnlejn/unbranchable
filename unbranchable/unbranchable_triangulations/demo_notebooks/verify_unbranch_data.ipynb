```python
import snappy, regina
from modules.branch_dfs import find_z2_merge_structures_dfs, has_z2_merge_structure_dfs
from tqdm.notebook import tqdm
from pathlib import Path
```

```python
with open(Path("../data/no_branch_up_to_11_tets.txt")) as f:
    no_branch = dict(line.strip().split(maxsplit=1) for line in f if line.strip())
print(len(no_branch))
```

```python
# the first promise is that the sig I gave you is actually the manifold I told you it was:
for name in tqdm(no_branch):
    # snappy 11 tets is not yet implemented. So you need the file locally from Shana Li's github.
    # In the meantime, i tot is easy check up to ten tets:
    if name[0:3]!='o11': # the pre 11 tet census sigs. Augment for o10 if you have not added SnapPy's 10 tet census
        M=snappy.Manifold(name)
        sig = no_branch[name]
        tri=regina.Triangulation3(sig)
        N=snappy.Manifold(tri)
        manifold_matches = False
        while not manifold_matches:
            N.randomize()
            N.simplify()
            try:
                if M.is_isometric_to(N):
                    manifold_matches=True
            except:
                pass # just got unlucky, reroll until we see the manifold
```

```python
# now the next promise is unbranchability.
# the first promise is that the sig I gave you is actually the manifold I told you it was:
for name in tqdm(no_branch):
    # This one should take about 30 minutes
    M=snappy.Manifold(name)
    sig = no_branch[name]
    tri=regina.Triangulation3(sig)
    assert not has_z2_merge_structure_dfs(tri)
    
```

```python
# now the uncoorientabel branchable subset is similar.
with open(Path("../data/no_coor_branch_up_to_11_tets.txt")) as f:
    no_coor_branch = dict(line.strip().split(maxsplit=1) for line in f if line.strip())
print(len(no_coor_branch))
```

```python
# now the next promise is unbranchability.
# the first promise is that the sig I gave you is actually the manifold I told you it was:
for name in tqdm(no_coor_branch):
    # This one should take a few hours.
    M=snappy.Manifold(name)
    sig = no_coor_branch[name]
    tri=regina.Triangulation3(sig)
    tri.orient() # must be oriented!
    assert not has_z2_merge_structure_dfs(tri,transverse=True)
```
